# Adding Images to Jupyter Notebooks
***
There are several ways to add an image to a Jupyter notebook.

## Option 1: Drag & Drop

If you drag an image over a **markdown** cell in a notebook, it will produce the required code for the image.

The following cell is the code that is produced when you drag the image file to the notebook and the next cell is the image that is produced when the cell is executed.  

The "\[\]\(\)" syntax is standard Markdown syntax for a hyperlink where "[]" is the name of the link and "()" is the web address. The "!" in front of the square brackets instructs Jupyter to render the image directly instead of making a typical hyperlink.

## Option 2: IPython

The IPython package has a function for displaying images.   This requires you to import the package as shown below. 

There are several optional arguments like: width, height, etc. that can be added in the call to Image() to change the appearance of the image that is displayed. 

```python
from IPython.display import Image
Image('images/saturn.jpg')
```

## Option 3: HTML

Option 3 is very similar to Option 1 except that it uses HTML syntax rather than Markdown.

<img src="images/saturn.jpg">

<img src="images/saturn.jpg">

# Rotating Images

Only the IPython approach gives you the ability to rotate an image in the notebook.  The syntax is slightly more involved and requires a second package, PIL. 

```python
from IPython.display import Image, display
from PIL import Image
img = Image.open("images/saturn.jpg")
rotated = img.rotate(-90, expand=True)
display(rotated)
```

## HTML Image Properties

There are many things you can do once you've imported an image, especially if you know some CSS and HTML! :) But if you don't, here are some useful tricks. 

1. **Changing the width** <br> 
    `<img src="path" `* width=100%*`>`
    * The width can be set in different units: %, pixels, and many others. 
    * Try setting the width of images above to 50% and to 50px, and see what happens!


2. **Wrapping text around the image** <br>
    `<img src="path" `* style="float:right"*`>`
    * You can set `float=left` to move the image all the way to the right, and make the text float on the left!
    
    
3. **Rotating the image** <br>
    Unfortunately, there is no easy way to rotate the image with HTML in Jupyter. :( <br>
    The easiest thing to do is to open the image in a viewer, rotate the image, and save the new rotated image.
